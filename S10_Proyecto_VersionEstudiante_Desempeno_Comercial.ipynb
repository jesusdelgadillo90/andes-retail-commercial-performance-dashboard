### Narrativa del Dashboard (Modelo SCQA)
Se describe de forma estructurada la historia completa del dashboard usando SCQA, incluyendo contexto, problemas detectados, preguntas de negocio y recomendaciones basadas en datos.

#### 🖥️ Vista General (Overview)

**S (Situación):**  
Entre 2024 y 2025, Andes Retail Group acumuló una facturación total de $5.53M USD a través de 5,000 pedidos distribuidos en Colombia, Perú y Chile.

**C (Complicación):**  
Al evaluar la evolución general, se evidencia una desaceleración comercial sostenida:Caída en la demanda e ingresos: Los ingresos disminuyeron un 6.8% en 2025 (2.67M) frente a 2024 (2.86M), impulsados principalmente por una reducción del 3.9% en el volumen total de pedidos (de 2,550 a 2,450 transacciones) y una menor base de clientes activos.                  

Desequilibrio geográfico: Colombia muestra un rezago significativo al aportar solo el 24.3% de la facturación global (1.35M) frente a Perú (2.16M) y Chile (2.03M).

**Q (Pregunta):**   
¿Qué decisiones estratégicas a nivel macro deben tomarse para revertir la caída en el volumen de pedidos de 2025 y nivelar el rendimiento comercial en los tres países?

**A (Respuesta):**  

Estrategia de Retención y Frecuencia de Compra: Diseñar programas de fidelización y campañas de reactivación (re-engagement) dirigidos a la base existente de clientes para recuperar la frecuencia y el volumen de pedidos de 2024.   

Plan de Penetración en Colombia: Fortalecer la presencia de marca y la cobertura comercial en Colombia para llevar sus niveles de facturación a la par con Perú y Chile.   

Incentivos por Volumen (Bundling): Ofrecer paquetes de productos que recompensen las compras de mayor volumen por transacción, atacando simultáneamente la recurrencia y la escala del pedido.

#### 🔎 Vista Detalle

**S (Situación):**  
La Vista Detalle analiza de forma granular la composición de los ingresos y la rentabilidad de las 5,000 transacciones, permitiendo evaluar el margen bruto promedio del 35.1% (1.94M USD en ganancia) según categorías de productos, temporadas del año, niveles de venta y segmentos de clientes

**C (Complicación):**  
El análisis en profundidad revela los tres factores específicos que explican la vulnerabilidad del negocio:Alta dependencia y brecha estacional: El 40.5% de los ingresos totales (2.24M) se concentran exclusivamente en Verano, mientras que en Invierno la facturación cae drásticamente a solo el 11.8% (653K).   
Desequilibrio de aportación en categorías: Deportes (1.44M) y Electrónica (1.41M) lideran la generación de margen, mientras que Ropa (1.32M) aporta el menor nivel de ingresos y ganancia sin presentar diferencias significativas en volumen de venta.   
Concentración de valor por segmento: El segmento Premium sostiene la mayor parte de la rentabilidad (2.60M en ingresos), mientras que los segmentos Estándar y Económico presentan una baja aportación relativa al margen total ($497K en el segmento Económico).

**Q (Pregunta):**  
¿Cómo se pueden aplanar los baches de facturación en Invierno, potenciar la rentabilidad de categorías rezagadas como Ropa y migrar clientes de segmentos bajos hacia mayor valor (Upselling), protegiendo el margen bruto del 35.1%?

**A (Respuesta):**  
Estrategia Comercial Anticíclica para Invierno: Lanzar ofertas dinámicas y promociones cruzadas (bundling) en las categorías de alto margen (Deportes y Electrónica) durante los meses de invierno para mitigar la caída estacional de ingresos.   

Optimización de Portafolio en Ropa y Hogar: Ajustar el mix de productos y la estrategia de pricing en Ropa y Hogar para incentivar un mayor volumen de compra fuera de la temporada alta de Verano.

Estrategia de Migración de Clientes (Upselling): Diseñar incentivos de compra dirigidos a los segmentos Estándar y Económico para elevar su frecuencia de compra, impulsar su migración hacia productos de mayor margen y diversificar la base de ingresos.
